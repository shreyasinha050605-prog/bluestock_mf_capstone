```python
import pandas as pd

fund_master = pd.read_csv(
    "/Users/shreyasinha/Desktop/bluestock_mf_capstone/data/raw/01_fund_master.csv"
)

print("="*50)
print("UNIQUE FUND HOUSES")
print("="*50)
print(fund_master["fund_house"].unique())

print("\n")

print("="*50)
print("UNIQUE CATEGORIES")
print("="*50)
print(fund_master["category"].unique())

print("\n")

print("="*50)
print("UNIQUE SUB CATEGORIES")
print("="*50)
print(fund_master["sub_category"].unique())

print("\n")

print("="*50)
print("UNIQUE RISK CATEGORIES")
print("="*50)
print(fund_master["risk_category"].unique())
```

```python
print("Number of Fund Houses:",
      fund_master["fund_house"].nunique())

print("Number of Categories:",
      fund_master["category"].nunique())

print("Number of Sub Categories:",
      fund_master["sub_category"].nunique())

print("Number of Risk Categories:",
      fund_master["risk_category"].nunique())
```

```python
print("\nFund Houses:")
for item in sorted(fund_master["fund_house"].unique()):
    print("-", item)

print("\nCategories:")
for item in sorted(fund_master["category"].unique()):
    print("-", item)

print("\nSub Categories:")
for item in sorted(fund_master["sub_category"].unique()):
    print("-", item)

print("\nRisk Categories:")
for item in sorted(fund_master["risk_category"].unique()):
    print("-", item)
```
