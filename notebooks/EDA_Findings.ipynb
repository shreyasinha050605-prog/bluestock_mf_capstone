# Exploratory Data Analysis (EDA) Findings

## Mutual Fund Industry Analysis (2022–2025)

This notebook summarizes the key insights identified during the Exploratory Data Analysis (EDA) phase of the Mutual Fund Capstone Project.

# Key EDA Findings

### 1. Strong Overall NAV Growth
The average NAV across all 40 mutual fund schemes increased steadily from approximately ₹210–220 in 2022 to more than ₹350 by 2026, indicating sustained growth in fund valuations.

### 2. Significant Market Rally in 2023
Average NAV levels recorded a notable increase during 2023, with growth of approximately 14.1%, reflecting improved market sentiment and stronger fund performance.

### 3. Continued Growth Momentum in 2024
The mutual fund industry maintained positive momentum during 2024, with average NAV increasing by approximately 12.5% compared to the previous year.

### 4. SBI Mutual Fund Remained the Largest AMC
SBI Mutual Fund consistently maintained the highest Assets Under Management (AUM) among all fund houses and remained the dominant player throughout the analysis period.

### 5. SIP Inflows Reached Record Levels
Monthly SIP inflows increased significantly from around ₹11,500 crore in early 2022 to more than ₹31,000 crore by December 2025, demonstrating rising retail investor participation.

### 6. Liquid Funds Attracted the Highest Net Inflows
Category-wise inflow analysis showed that Liquid Funds consistently received the highest inflows, indicating strong investor preference for liquidity and lower-risk investment options.

### 7. Young Investors Dominated the Investor Base
The 26–35 age group represented the largest share of investors, accounting for more than 40% of the total investor population.

### 8. T30 Cities Contributed the Majority of Investments
Approximately 66% of total investments originated from T30 cities, highlighting the continued dominance of major urban centers in mutual fund participation.

### 9. Folio Count Nearly Doubled
Total mutual fund folios increased from approximately 13.3 crore in 2022 to 26.1 crore in 2025, indicating substantial growth in investor participation and market penetration.

### 10. Positive Correlation Observed Among Most Funds
Correlation analysis revealed that most selected mutual fund schemes exhibited moderate to strong positive correlations, indicating similar responses to broader market movements.

# Conclusion

The exploratory data analysis reveals strong expansion in the Indian mutual fund industry between 2022 and 2025. Growth was observed across NAV levels, AUM, SIP inflows, and folio counts. Increasing retail participation, rising SIP contributions, and sustained industry leadership by major AMCs such as SBI Mutual Fund indicate a healthy and rapidly growing mutual fund ecosystem.
